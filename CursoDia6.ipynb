- Una clase NO hereda de otra, sino que contiene un objeto de esa otra clase.

## 9.- Composicion vs Herencia
Esto se va a responder con pura: Teoria

No significa que alguna sea mejor que otra simplemente que aveces una te puede convenir mas que otra:

Herencia: un gato ES un animal (no "Un gato TIENE un animal")

Composicion: Una persona TIENE una cuenta (no "Una persona Es un Banco")

```python
# EJERCICIO: Aplicando Herencia, Composición y Polimorfismo

class Vehiculo:
    """Clase base para todos los vehículos. Define la interfaz 'mover'."""
    def __init__(self, marca):
        self.marca = marca

    def mover(self):
        # Este método se define en las clases hijas (Polimorfismo)
        raise NotImplementedError("Este método DEBE ser implementado por la subclase.")


class Motor:
    """Clase que representa un motor (usada en Composición)."""
    def __init__(self, caballos):
        self.caballos = caballos

    def potencia(self):
        return f"{self.caballos} HP"

# ----------------------------------------
# Clases Hijas con Composición y Polimorfismo
# ----------------------------------------

class Coche(Vehiculo):
    """Coche: usa Motor por Composición."""
    def __init__(self, marca, caballos):
        super().__init__(marca)
        # 🚗 COMPOSICIÓN: El Coche tiene un Motor.
        self.motor = Motor(caballos)

    def mover(self):
        # 💨 POLIMORFISMO: Implementación específica de mover para Coche.
        return f"El coche {self.marca} avanza con motor de {self.motor.potencia()}"


class Moto(Vehiculo):
    """Moto: usa Motor por Composición."""
    def __init__(self, marca, caballos):
        super().__init__(marca)
        # 🏍️ COMPOSICIÓN: La Moto tiene un Motor.
        self.motor = Motor(caballos)

    def mover(self):
        # 💨 POLIMORFISMO: Implementación específica de mover para Moto.
        return f"La moto {self.marca} acelera con motor de {self.motor.potencia()}"


class Bicicleta(Vehiculo):
    """Bicicleta: No necesita Motor, pero hereda de Vehiculo."""
    # Nota: No requiere __init__ si no añade atributos nuevos, pero necesita inicializar la marca.
    # Necesitas asegurar que Bicicleta("Trek") funciona, por lo que hereda __init__ de Vehiculo.

    def mover(self):
        # 💨 POLIMORFISMO: Implementación específica de mover para Bicicleta.
        return f"La bicicleta {self.marca} avanza con pedaleo humano"

# ----------------------------------------
# Función de Prueba (DEBE ir fuera de las clases)
# ----------------------------------------

def probar_vehiculos(lista):
    """Función para demostrar el Polimorfismo."""
    print("\n--- Demostración de Polimorfismo ---")
    for v in lista:
        # Llama a la versión correcta de 'mover()' según el tipo de objeto (Coche, Moto, Bicicleta)
        print(v.mover())

# --- Ejecución y Prueba ---

# 1. Creamos la lista de vehículos (distintas clases)
vehiculos = [
    Coche("Toyota", 150),
    Moto("Yamaha", 90),
    Bicicleta("Trek")
]

# 2. Llamamos a la función de prueba
probar_vehiculos(vehiculos)
```
